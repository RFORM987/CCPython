```python
# Conversion des données dans des formats plus exploitables

data['Product_data']['Standard Cost'] = data['Product_data']['Standard Cost'].str.replace("$","").str.replace(",","").astype(float)
data['Product_data']['List Price'] = data['Product_data']['List Price'].str.replace("$","").str.replace(",","").astype(float)
data['Sales_data']['Unit Price'] = data['Sales_data']['Unit Price'].str.replace("$","").str.replace(",","").astype(float)
data['Sales_data']['Extended Amount'] = data['Sales_data']['Extended Amount'].str.replace("$","").str.replace(",","").astype(float)
data['Sales_data']['Product Standard Cost'] = data['Sales_data']['Product Standard Cost'].str.replace("$","").str.replace(",","").astype(float)
data['Sales_data']['Total Product Cost'] = data['Sales_data']['Total Product Cost'].str.replace("$","").str.replace(",","").astype(float)
data['Sales_data']['Sales Amount'] = data['Sales_data']['Sales Amount'].str.replace("$","").str.replace(",","").astype(float)



print(data['Product_data']['Standard Cost'])
```

```python
# Exo 3 :

#Listage des champs vides dans les dataframes

for nom , dataframe in data.items():
    print(nom, " : " , dataframe.isna().sum())
```

```python
# Isolation des ventes invalides

sales = data['Sales_data']

invalid_sales = ~np.isclose(sales["Sales Amount"] , 
sales["Unit Price"] * sales["Order Quantity"] * (1 - sales["Unit Price Discount Pct"] / 100) ,
rtol=1e-9 ,
atol=1e-8
)

print(sales[["Unit Price","Order Quantity","Unit Price Discount Pct","Sales Amount"]][invalid_sales])
```

```python
# Calcul des Marges, Ratio et Discount

sales = data['Sales_data']

sales["Price Ratio"] = sales["Unit Price"] / sales["Product Standard Cost"]

salesbyproduct = pd.merge(sales,data["Product_data"],on="ProductKey")

sales["Margin"] = salesbyproduct["Sales Amount"] - salesbyproduct["Standard Cost"]* salesbyproduct["Order Quantity"]

sales["Unit Price Discount Pct"] = 100 - ((salesbyproduct["Standard Cost"] * salesbyproduct["Order Quantity"])/salesbyproduct["Sales Amount"]) * 100

print(sales["Unit Price Discount Pct"])

print(sales[sales["Price Ratio"] > 2.6738])

print(sales[["Sales Amount","Total Product Cost", "Margin"]])
```

```python
#Isolation du numéro de ligne dans une colonne a part

orders = data['Sales Order_data']

orders["Line number"] = orders["Sales Order Line"].str.split(" ").str[-1]

print(orders["Line number"])
```

```python
# Jour 2 :

# Exo 5 :

# Statistique par territoire puis schématisation simple

datasets = {
    val : d.copy()
    for val,d in pd.merge(data["Sales_data"],data["Sales Territory_data"], on="SalesTerritoryKey").groupby("SalesTerritoryKey")
}

dic = {}

for key, dataset in datasets.items() : 
    print(key)
    print(dataset["Country"].iloc[0])
    print(dataset["Region"].iloc[0])
    print("moyenne :",dataset["Sales Amount"].mean())
    print("mediane :" ,dataset["Sales Amount"].median())
    print("écart-type:",dataset["Sales Amount"].std())
    print("max :",dataset["Sales Amount"].max())
    print("min :",dataset["Sales Amount"].min())
    print("premier quartile :",dataset["Sales Amount"].quantile(0.25))
    print("troisime quartile :",dataset["Sales Amount"].quantile(0.75))
    dic[dataset["Country"].iloc[0]+dataset["Region"].iloc[0]] = dataset["Sales Amount"].sum()
    print("\n\n")

print(pd.Series(dic).plot(kind='bar'))
```

```python
# Statistique par produit puis schématisation simple

datasets = {
    val : d.copy()
    for val,d in pd.merge(data["Sales_data"],data["Product_data"], on="ProductKey").groupby("ProductKey")
}

dic = {}

for key, dataset in datasets.items() : 
    print(key)
    print(dataset["Model"].iloc[0])
    print(dataset["Color"].iloc[0])
    print("moyenne :",dataset["Sales Amount"].mean())
    print("mediane :" ,dataset["Sales Amount"].median())
    print("écart-type:",dataset["Sales Amount"].std())
    print("max :",dataset["Sales Amount"].max())
    print("min :",dataset["Sales Amount"].min())
    print("premier quartile :",dataset["Sales Amount"].quantile(0.25))
    print("troisime quartile :",dataset["Sales Amount"].quantile(0.75))
    dic[dataset["Model"].iloc[0]] = dataset["Sales Amount"].sum()
    print("\n\n")

print(pd.Series(dic).nlargest(20).plot(kind='bar'))
```

```python
# Exo 7 :

# Création de série temporelle et affichage par graphique a bougies

salesbyorderdateM = pd.merge(data["Sales_data"], data["Date_data"],left_on="OrderDateKey",right_on="DateKey").groupby(pd.Grouper(key="Date", freq="1ME"))["Sales Amount"].ohlc()

print(salesbyorderdateM)

salesbyorderdateTri = salesbyorderdateM.resample("3ME").ohlc()

print(salesbyorderdateTri)

salesbyorderdateY = salesbyorderdateM.resample("1YE").ohlc()

print(salesbyorderdateY)

a, graph = plot.subplots(figsize=(12,6))

for date , entry in salesbyorderdateM.iterrows():
    x = matdat.date2num(date)
    open = entry["open"]
    high = entry["high"]
    low = entry["low"]
    close = entry["close"]

    couleur = "green" if close>=open else "red"

    graph.plot([x,x],[low,high],color=couleur)

    graph.add_patch(Rectangle((x-5,min(open,close)),10,max(abs(close-open),1),facecolor=couleur))

graph.xaxis_date()
graph.xaxis.set_major_formatter(matdat.DateFormatter("%m/%Y"))


plot.show()

```

```python
#Exo 8 :

# Création de Fullmerge, un merge global de toutes les données

print(data["Reseller_data"])

Fullmerge = data["Sales_data"].merge( data["Date_data"],
    left_on="OrderDateKey",right_on="DateKey") \
    .merge(data["Customer_data"], on="CustomerKey") \
    .merge(data["Reseller_data"], on="ResellerKey") \
    .merge(data["Sales Territory_data"], on="SalesTerritoryKey") \
    .merge(data["Sales Order_data"], on="SalesOrderLineKey") \
    .merge(data["Product_data"], on="ProductKey") 

print(Fullmerge[["Unit Price","Extended Amount","Margin","Product Standard Cost","Sales Amount"]])
    
```

```python
# Calcul des ventes par produit et territoire

crossProductTerritory = pd.crosstab(Fullmerge["SalesTerritoryKey"],Fullmerge["ProductKey"], values=Fullmerge["Sales Amount"], aggfunc="sum").fillna(-1)

print(crossProductTerritory)

# Calcul du nombre de produits par modèle

productsbymodel = data["Product_data"].groupby(["Model"])["ProductKey"].count()

print(productsbymodel)

# Calcul du nombre de commandes par client

clientmaxCommands= Fullmerge.groupby(["CustomerKey"])["SalesOrderLineKey"].count().sort_values(ascending=False)
clientmaxCommands = clientmaxCommands.drop(-1)
print(clientmaxCommands)

# Calcul des dépenses par client

clientmaxSpent= Fullmerge.groupby(["CustomerKey"])["Sales Amount"].sum().sort_values(ascending=False)
clientmaxSpent = clientmaxSpent.drop(-1)
print(clientmaxSpent)

# Calcul des ventes par produit et par model

crossProductCatTerritory = pd.crosstab(Fullmerge["SalesTerritoryKey"],Fullmerge["Model"], values=Fullmerge["Sales Amount"], aggfunc="sum").fillna(-1)

print(crossProductCatTerritory)
```

```python
# Exo 9 :

# Calcul des ventes des revendeurs

perfRevendeur = Fullmerge.groupby(["ResellerKey"])["Sales Amount"].sum().sort_values(ascending=False)
perfRevendeur = perfRevendeur.drop(-1)
print(perfRevendeur)

# Calcul des marges des revendeurs

margRevendeur = Fullmerge.groupby(["ResellerKey"])["Margin"].sum().sort_values(ascending=False)
margRevendeur = margRevendeur.drop(-1)
print(margRevendeur)

# Merge des précédents et affichage dans un graphique ( on divise les ventes par 10 pour plus de lisibilité )

revmargtosales = pd.merge(perfRevendeur,margRevendeur, on="ResellerKey")

print(revmargtosales)

revmargtosales["Sales Amount"] = revmargtosales["Sales Amount"]/10

revmargtosales = pd.merge(revmargtosales,data["Reseller_data"],on="ResellerKey")[["Reseller","Sales Amount","Margin"]]

print(revmargtosales)

graph = revmargtosales.head(30).plot(x="Reseller",kind='bar',figsize=(20,8),rot=50)
graph.set_title("Ventes et marges par revendeur")
plot.setp(graph.get_xticklabels(),ha='right')
plot.savefig("sales_and_margins_by_reseller.png")
plot.show()

```

```python
# Reproduction de la cellules précédente en replacant les revendeurs par les territoires

perfTerritoire = Fullmerge.groupby(["SalesTerritoryKey"])["Sales Amount"].sum().sort_values(ascending=False)
print(perfTerritoire)

margTerritoire = Fullmerge.groupby(["SalesTerritoryKey"])["Margin"].sum().sort_values(ascending=False)
print(margTerritoire)

revmargtosales = pd.merge(perfTerritoire,margTerritoire, on="SalesTerritoryKey")

print(revmargtosales)

revmargtosales["Sales Amount"] = revmargtosales["Sales Amount"]/10

revmargtosales = pd.merge(revmargtosales,data["Sales Territory_data"],on="SalesTerritoryKey")[["Region","Country","Sales Amount","Margin"]]

revmargtosales["Territory"] = revmargtosales["Region"]+" "+revmargtosales["Country"]

print(revmargtosales)

graph = revmargtosales.head(30).plot(x="Territory",kind='bar',figsize=(20,8),rot=50)
graph.set_title("Ventes et marges par Territoire")
plot.setp(graph.get_xticklabels(),ha='right')
plot.savefig("sales_and_margins_by_reseller.png")
plot.show()
```
